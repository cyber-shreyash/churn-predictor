```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    X_train_scaled,
    y_train,
    cv=20,
    scoring='accuracy'
)

print("CV Scores:", scores)
print("Mean Accuracy:", scores.mean())
```

```python
import optuna

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold, cross_val_score


def objective_lr(trial):

    C = trial.suggest_float("C", 0.001, 10, log=True)

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("lr", LogisticRegression(
            C=C,
            class_weight='balanced',
            max_iter=1000
        ))
    ])

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring='roc_auc'
    )

    return scores.mean()


study_lr = optuna.create_study(direction="maximize")

study_lr.optimize(objective_lr, n_trials=30)

print("Best LR parameters:", study_lr.best_params)
print("Best CV ROC-AUC:", study_lr.best_value)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Get churn probabilities from the final model
y_prob = model_lr.predict_proba(X_test)[:, 1]

# Try different thresholds
thresholds = np.arange(0.30, 0.71, 0.05)

results = []

for threshold in thresholds:

    y_pred = (y_prob >= threshold).astype(int)

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred)
    })

# Display results
import pandas as pd

results_df = pd.DataFrame(results)

print(results_df)

# Find threshold with best F1-score
best = results_df.loc[results_df["F1"].idxmax()]

print("\nBest Threshold:", best["Threshold"])
print("Best Accuracy:", best["Accuracy"])
print("Best Precision:", best["Precision"])
print("Best Recall:", best["Recall"])
print("Best F1:", best["F1"])
```
